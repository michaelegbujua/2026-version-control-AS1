5. Write a program that finds the most common element in a list and how many times it appears.

```python
numbers = [10,20,10,30,20,10,40,30,10]
unique_num = set(numbers)
#print(unique_num)

most_common = None
highest_count = 0

for i in unique_num:
    count=numbers.count(i)
    #print(count)
    if count > highest_count:
        highest_count = count
        most_common = i
        
print(f"Most common element is : {most_common}, frequency of appearance: {highest_count}")
```

6. Create a program that counts the number of vowels and consonants in a sentence.

```python
vowels = ['a','e','i','o','u']
consonants =[]

count_v=0
count_c=0
sentence = input('enter a short sentence: ').lower()
for char in sentence:
    #remove empty spaces
    if char.isalpha():
        if char in vowels:
            count_v +=1
        else:
            consonants.append(char)
            count_c +=1
        
print(f"the vowel count is : {count_v}")
print(f"the consonant count is : {count_c}")
        
```

9. Create a program that sums user-entered numbers until the total exceeds 100, 
then stops and reports how many numbers were entered.


```python
num=[]
sum_num=0
while sum_num<100:
    val= int(input('enter a value:'))
    num.append(val)
    sum_num= sum(num)
print(f" total sum: {sum_num}")
print(f"{len(num)} numbers were entered")

    
```

12. Create a program that prints the Fibonacci sequence up to a specific count provided by the user.

```python
#fibonacci sequence always add with the last value behind it
 
count= int(input('How may Fibonacci numbers would you like ?'))
a,b = 0,1
fib_sequence = []
for i in range(count):
    fib_sequence.append(a)
    a,b = b, a+b
    
print(f"Fibonacci sequence: {fib_sequence}")
```
