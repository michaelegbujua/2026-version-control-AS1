9.Create a function count_vowels that takes a string and returns the number of vowels in it.

```python
def count_vowel(string_val):
    string_val = str(string_val).lower()
    vowel = ['a','e','i','o','u']
    count = 0
    
    for i in string_val:
        
        if i in vowel:
            count =count + 1
    
    if count ==0:
        return f"no vowel in {string_val}"
    return count
    
    
```

```python
count_vowel('Accommodation')
```

10.	Define a function find_longest_word that takes a list of words and returns the longest one.

```python
def longest_word(words):
    longest=""
    for word in words:
        if len(word) > len(longest):
            longest = word
    return longest
```

```python
longest_word(["Life",  "is", "precious"])
```

11.	Write a function grade_calculator that takes a score (0-100) and returns the letter grade (A, B, C, D,E, F).

```python
def grade_calculator(score):
    if score in range(70,101):
        return 'A'
    elif score in range(60,70):
        return 'B'
    elif score in range(50,60):
        return 'C'
    elif score in range(45,50):
        return 'D'
    elif score in range(40,45):
        return 'E'
    elif score>=0 and score<=39:
        return 'F'
    else:
        return "Invalid Score, enter score between 0-100"
```

```python
grade_calculator(800)
```

12.	Write a function file_extension that extracts the file extension from a filename, handling cases where there might be no extension.

```python
def file_extension(filename):
    if "." in filename:
        return filename.rsplit(".", 1)[1]  # split on dot, take last part
    else:
        return "no extension"
```

```python
file_extension('manual.docx')
print(file_extension("archive"))       # Output: no extension
print(file_extension("data.backup.tar")) 
```

13.	Write a function safe_get_item that tries to access a list item by index and returns None if the index is out of bounds.

```python
def safe_get_item(lst, index):
    if 0<= index<len(lst):
        return lst[index]
    else:
        return None
```

```python

safe_get_item('name', 5)
```

14.	Create a function clean_data that takes a list of mixed data types and returns only the numeric values as a new list

```python
def clean_data(data):
    numeric_values= []
    for item in data:
        if isinstance(item, (int, float)) and not isinstance(item, bool):
            numeric_values.append(item)
    return numeric_values
```

```python
clean_data([3,22,7,9])
```
