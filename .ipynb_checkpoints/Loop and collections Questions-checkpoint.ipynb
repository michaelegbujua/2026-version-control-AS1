1. Write a program that counts down from 10 to 1, then prints "Blast off!"

```python
count = 10
for i in range(0,11):
    print(count)
    count-=1
    
    if count<1:
        print("Blast Off!")
        break
    
```

2. Write a program that finds all numbers between 1 and 1000 that are divisible by 7.

```python
#using  a list and converting it to a set
num_list=[]
for i in range(1, 1000):
    num=i%7
    num_list.append(num)
num_set= set(num_list) #to avoid repetition of numbers convert to set
    
print(num_set)
```

3. Given a list of numbers, find and print the largest and smallest values.


```python
list_of_num = [3,9,4,5,6,87, 3, 1, 9]
list_of_num.sort()
print(f"the largest value: {list_of_num[-1]} and the smallest value: {list_of_num[0]}")
```

4. Write a program that removes all duplicate values from a list and prints the unique elements.

```python
list_val = [3,5,3,9,10,4,5,8, 10,10,12,7,22,13]
dupl_val = list(set(list_val))
print(dupl_val)
```
